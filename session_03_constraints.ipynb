## Worked example

Color a small map with three colors. After assigning a region, remove incompatible choices from adjacent regions. Show one backtrack.

**Discussion question:** CSP ingredients?

## Map-coloring model

Two neighboring regions cannot use the same color.

```python
ADJ={"WA":["NT","SA"],"NT":["WA","SA","Q"],"SA":["WA","NT","Q","NSW","V"],"Q":["NT","SA","NSW"],"NSW":["SA","Q","V"],"V":["SA","NSW"],"T":[]}
COLORS=("red","green","blue")
def available(v,a):
    return [c for c in COLORS if all(a.get(n)!=c for n in ADJ[v])]
print("First domain:",available("WA",{}))
```

## Backtracking plus MRV

Count recursive calls, not wall-clock time on a tiny instance.

```python
def solve(mrv=True):
    calls=[0]
    def search(a):
        calls[0]+=1
        if len(a)==len(ADJ):return dict(a)
        rest=[v for v in ADJ if v not in a]
        v=min(rest,key=lambda x:len(available(x,a))) if mrv else rest[0]
        for color in available(v,a):
            a[v]=color;result=search(a)
            if result is not None:return result
            del a[v]
        return None
    return search({}),calls[0]
assignment,steps=solve()
print(assignment,"calls:",steps,"fixed calls:",solve(False)[1])
assert all(assignment[v]!=assignment[n] for v in ADJ for n in ADJ[v])
```

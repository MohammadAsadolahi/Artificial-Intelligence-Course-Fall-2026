## Worked example

Given facts enrolled, paid, course_open, and rules enrolled AND paid -> eligible; eligible AND course_open -> can_register, derive the conclusion and justify each step.

**Discussion question:** What is forward chaining?

## Rules and initial knowledge

Represent premises as sets and conclusions as strings.

```python
facts={"enrolled","paid","course_open"}
rules=[({"enrolled","paid"},"eligible"),({"eligible","course_open"},"can_register"),({"can_register"},"show_button")]
print("Initial facts:",facts)
```

## Forward-chaining engine

Keep a trace of each conclusion's premises.

```python
def infer(initial,rules):
    known=set(initial);proof={}
    changed=True
    while changed:
        changed=False
        for before,after in rules:
            if before<=known and after not in known:
                known.add(after);proof[after]=sorted(before);changed=True
    return known,proof
known,proof=infer(facts,rules)
print("All facts:",sorted(known),"Proof:",proof)
assert "can_register" in known
limited,_=infer(facts-{"paid"},rules)
assert "can_register" not in limited
```

## مثال حل‌شده

در دنیای جاروبرقی دو اتاقه، عامل می‌داند در کدام اتاق قرار دارد و آیا همان اتاق کثیف است یا خیر. یک سیاست تصمیم‌گیری تعریف کنید: اگر اتاق فعلی کثیف بود، `CLEAN` (تمیز کردن) و در غیر این صورت حرکت به اتاق دیگر (`RIGHT` یا `LEFT`). چهار اقدام پیاپی را دنبال کنید و نتیجه را با حالتی مقایسه کنید که عامل هیچ کاری انجام نمی‌دهد (`WAIT`).

**پرسش کلاسی:** PEAS مخفف چه مؤلفه‌هایی است و هر مؤلفه در این مثال به چه چیزی اشاره دارد؟

## مدل‌سازی محیط و سیاست انتخاب اقدام

وضعیت محیط را به‌صورت صریح نمایش دهید و یک سیاست تصمیم‌گیری قطعی (Deterministic) پیاده‌سازی کنید.

دو اتاق `A` و `B` داریم؛ در حالت آغازین هر دو کثیف هستند و عامل در اتاق `A` قرار دارد. تابع `agent` اقدام بعدی را انتخاب می‌کند و تابع `step` وضعیت جدید محیط را برمی‌گرداند.

```python
from copy import deepcopy
START={"position":"A","dirty":{"A":True,"B":True}}
def agent(w):
    return "CLEAN" if w["dirty"][w["position"]] else ("RIGHT" if w["position"]=="A" else "LEFT")
def step(w,action):
    w=deepcopy(w)
    if action=="CLEAN": w["dirty"][w["position"]]=False
    elif action=="RIGHT": w["position"]="B"
    elif action=="LEFT": w["position"]="A"
    return w
print("نخستین اقدام:",agent(START))
```

## ارزیابی با خط پایه (Baseline)

برای مقایسهٔ منصفانه، وضعیت اولیه را در هر دو سیاست یکسان نگه دارید.

- **عامل مبتنی بر قانون:** اگر اتاق فعلی کثیف باشد، آن را تمیز می‌کند و در غیر این صورت به اتاق دیگر می‌رود.
- **خط پایه:** همیشه `WAIT` را انتخاب می‌کند و هیچ تغییری در وضعیت محیط ایجاد نمی‌کند.

در هر مرحله، تعداد اتاق‌هایی را که هنوز کثیف‌اند ثبت و دو نتیجه را با هم مقایسه کنید.

```python
def episode(policy,start,steps=4):
    w=deepcopy(start); history=[]
    for _ in range(steps):
        action=policy(w);w=step(w,action)
        history.append((action,sum(w["dirty"].values())))
    return w,history
smart,trace=episode(agent,START)
idle,other=episode(lambda w:"WAIT",START)
print("عامل:",trace,"خط پایه:",other)
assert sum(smart["dirty"].values())<sum(idle["dirty"].values())
```

## تمرین

وضعیت آغازین را طوری تغییر دهید که یکی از اتاق‌ها از ابتدا تمیز باشد. گزارش کنید عامل در چهار مرحله چه اقدام‌هایی انجام می‌دهد و در پایان چند اتاق کثیف باقی می‌ماند.

**تکلیف:** حسگری را همراه با نویز به مدل اضافه کنید و نتیجهٔ ۱۰۰ وضعیت آغازینِ قابل بازتولید را میانگین بگیرید.

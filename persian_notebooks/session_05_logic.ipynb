### اهداف یادگیری
- مفهوم **حقیقت (Fact)** و **قواعد هورن (Horn Rules)** را توضیح دهید و مثال بزنید.

- روش **زنجیره‌سازی پیشرو (Forward Chaining)** را پیاده‌سازی و مراحل استنتاج را دنبال کنید.
- **محدودیت‌های استدلال نمادین** را بشناسید و موقعیت‌هایی را که به عدم قطعیت نیاز دارند، تشخیص دهید.

ابتدا **README.md** اصلی دوره را بخوانید. پیش از اجرای هر سلول، خروجی آن را حدس بزنید.

## مثال حل‌شده

فرض کنید پایگاه دانش دانشگاه سه حقیقت اولیه دارد:

- `enrolled`: دانشجو ثبت‌نام اولیه را انجام داده است.
- `paid`: شهریه پرداخت شده است.
- `course_open`: درس برای ثبت‌نام باز است.

دو قاعده داریم:

1. اگر `enrolled` و `paid` برقرار باشند، نتیجهٔ `eligible` (واجد شرایط بودن) به دست می‌آید.
2. اگر `eligible` و `course_open` برقرار باشند، نتیجهٔ `can_register` (امکان ثبت‌نام) به دست می‌آید.

نتیجه را مرحله‌به‌مرحله استخراج کنید و برای هر گام مشخص کنید کدام حقایق و قواعد آن را توجیه می‌کنند.

**پرسش کلاسی:** زنجیره‌سازی پیشرو (Forward Chaining) چیست؟

## قواعد و دانش اولیه

مقدمات هر قاعده را با یک `set` و نتیجهٔ آن را با یک رشته نمایش می‌دهیم.

- `facts` شامل حقایقی است که در آغاز می‌دانیم.
- هر عضو `rules` یک جفت `(premises, conclusion)` است.
- قاعدهٔ سوم نشان می‌دهد که از `can_register` می‌توان به `show_button` (نمایش دکمهٔ ثبت‌نام) رسید.

در این بخش فقط دانش اولیه و قواعد را تعریف می‌کنیم؛ هنوز استنتاج انجام نشده است.

```python
facts={"enrolled","paid","course_open"}
rules=[({"enrolled","paid"},"eligible"),({"eligible","course_open"},"can_register"),({"can_register"},"show_button")]
print("حقایق اولیه:",facts)
```

## موتور زنجیره‌سازی پیشرو

تابع `infer` از مجموعهٔ حقایق اولیه شروع می‌کند و تا زمانی که قاعده‌ای حقیقت جدیدی تولید کند، به اجرای قواعد ادامه می‌دهد.

- `known` مجموعهٔ حقایق شناخته‌شده است.
- `proof` مقدماتی را نگه می‌دارد که هر نتیجهٔ تازه را توجیه می‌کنند.
- وقتی دیگر حقیقت جدیدی به دست نیاید، الگوریتم متوقف می‌شود.

ابتدا بررسی کنید که آیا `can_register` نتیجه می‌شود. سپس حقیقت `paid` را حذف کنید و دوباره استنتاج را انجام دهید.

**نکته:** نداشتن مدرک برای اثبات یک گزاره، لزوماً به معنای نادرست بودن آن گزاره نیست.

```python
def infer(initial,rules):
    known=set(initial);proof={}
    changed=True
    while changed:
        changed=False
        for before,after in rules:
            if before<=known and after not in known:
                known.add(after);proof[after]=sorted(before);changed=True
    return known,proof
known,proof=infer(facts,rules)
print("همهٔ حقایق:",sorted(known),"مسیر استنتاج:",proof)
assert "can_register" in known
limited,_=infer(facts-{"paid"},rules)
assert "can_register" not in limited
```

## تمرین

حقیقت `paid` را از دانش اولیه حذف کنید و ببینید کدام نتیجه‌ها دیگر قابل استنتاج نیستند. توضیح دهید هر نتیجه به کدام مقدمات وابسته بوده است.

**تکلیف:** دو قاعدهٔ دیگر دربارهٔ ثبت‌نام دانشگاهی اضافه کنید و مسیر استنتاج هر نتیجه را ثبت و توضیح دهید.

**برداشت نادرست رایج:** نبودن یک حقیقت در پایگاه دانش، لزوماً به معنای نادرست بودن آن نیست.

نوت‌بوک را از یک کرنل تازه و به‌ترتیب اجرا کنید و دست‌کم یک محدودیت استدلال نمادین (مانند اطلاعات ناقص یا قواعد دارای استثنا) را توضیح دهید.

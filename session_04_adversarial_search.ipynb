## Worked example

Given two X moves and two O replies for each, calculate the minimum outcome per move, then choose the maximum of those minima.

**Discussion question:** Who chooses minimum nodes?

## Board rules

Represent a board with nine squares and all eight winning lines.

```python
LINES=[(0,1,2),(3,4,5),(6,7,8),(0,3,6),(1,4,7),(2,5,8),(0,4,8),(2,4,6)]
def result(b):
    for a,c,d in LINES:
        if b[a] and b[a]==b[c]==b[d]:return 1 if b[a]=="X" else -1
    return 0 if all(b) else None
def moves(b):return [i for i,x in enumerate(b) if not x]
def moved(b,i,p):
    out=b.copy();out[i]=p;return out
print("Empty-board legal moves:",moves([""]*9))
```

## Minimax

Scores are always defined from X's point of view.

```python
def minimax(b,turn):
    terminal=result(b)
    if terminal is not None:return terminal
    other="O" if turn=="X" else "X"
    values=[minimax(moved(b,i,turn),other) for i in moves(b)]
    return max(values) if turn=="X" else min(values)
board=["X","O","X","O","X","","","","O"]
scores={i:minimax(moved(board,i,"X"),"O") for i in moves(board)}
print("Scores:",scores)
assert max(scores,key=scores.get)==6 and scores[6]==1
```

## Alpha-beta

A correct optimization preserves the returned game value.

```python
def alphabeta(b,turn,alpha=-2,beta=2):
    t=result(b)
    if t is not None:return t
    if turn=="X":
        value=-2
        for i in moves(b):
            value=max(value,alphabeta(moved(b,i,"X"),"O",alpha,beta))
            alpha=max(alpha,value)
            if alpha>=beta:break
    else:
        value=2
        for i in moves(b):
            value=min(value,alphabeta(moved(b,i,"O"),"X",alpha,beta))
            beta=min(beta,value)
            if alpha>=beta:break
    return value
print("Alpha-beta:",alphabeta(board,"X"))
assert alphabeta(board,"X")==minimax(board,"X")
```

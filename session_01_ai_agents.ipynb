## Worked example

In a two-room vacuum world the agent sees its position and whether the current room is dirty. Write a policy: CLEAN if dirty, otherwise MOVE. Trace four actions and compare against doing nothing.

**Discussion question:** What does PEAS represent?

## World and action policy

Represent the world explicitly and implement a deterministic policy.

```python
from copy import deepcopy
START={"position":"A","dirty":{"A":True,"B":True}}
def agent(w):
    return "CLEAN" if w["dirty"][w["position"]] else ("RIGHT" if w["position"]=="A" else "LEFT")
def step(w,action):
    w=deepcopy(w)
    if action=="CLEAN": w["dirty"][w["position"]]=False
    elif action=="RIGHT": w["position"]="B"
    elif action=="LEFT": w["position"]="A"
    return w
print("First action:",agent(START))
```

## Evaluate a baseline

Keep the initial state identical across policies.

```python
def episode(policy,start,steps=4):
    w=deepcopy(start); history=[]
    for _ in range(steps):
        action=policy(w);w=step(w,action)
        history.append((action,sum(w["dirty"].values())))
    return w,history
smart,trace=episode(agent,START)
idle,other=episode(lambda w:"WAIT",START)
print("Agent:",trace,"baseline:",other)
assert sum(smart["dirty"].values())<sum(idle["dirty"].values())
```

## Worked example

On a 5×5 blocked grid, label start and goal, then assign g, h and f to frontier positions. Explain why the closest-looking cell is not always the cheapest path.

**Discussion question:** When is BFS shortest?

## Define grid and BFS

A valid successor remains in bounds and outside walls.

```python
from collections import deque
GRID=[".....",".###.","...#.",".#...","....."]
START,GOAL=(0,0),(4,4)
def neighbors(p):
    r,c=p
    for dr,dc in [(1,0),(-1,0),(0,1),(0,-1)]:
        nr,nc=r+dr,c+dc
        if 0<=nr<5 and 0<=nc<5 and GRID[nr][nc]==".":
            yield (nr,nc)
def path(prev,p):
    out=[p]
    while p in prev: p=prev[p];out.append(p)
    return out[::-1]
def bfs():
    q=deque([START]); seen={START};prev={};count=0
    while q:
        cur=q.popleft();count+=1
        if cur==GOAL:return path(prev,cur),count
        for nxt in neighbors(cur):
            if nxt not in seen: seen.add(nxt);prev[nxt]=cur;q.append(nxt)
    return None,count
print("BFS:",bfs())
```

## A* and controlled comparison

Use g-score bookkeeping to ignore stale queue entries.

```python
from heapq import heappush,heappop
def astar():
    h=lambda q:abs(q[0]-GOAL[0])+abs(q[1]-GOAL[1])
    heap=[(h(START),0,START)];best={START:0};prev={};count=0
    while heap:
        _,g,cur=heappop(heap)
        if g!=best[cur]:continue
        count+=1
        if cur==GOAL:return path(prev,cur),count
        for nxt in neighbors(cur):
            new=g+1
            if new<best.get(nxt,float("inf")):
                best[nxt]=new;prev[nxt]=cur;heappush(heap,(new+h(nxt),new,nxt))
    return None,count
b,be=bfs();a,ae=astar()
print("Path length:",len(a)-1,"BFS expanded:",be,"A* expanded:",ae)
assert b is not None and len(a)==len(b)
```
